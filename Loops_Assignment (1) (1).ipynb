## 🏨 8. Hotel Room Booking

### Goal:
    Manage simple room booking.

### Task:

    Create a list of rooms [101, 102, 103, 104, 105]
    
    Ask user to choose a room.
    
    If room already booked, print “Room not available.”
    
    Continue until all rooms are booked.

### Example:

    Available rooms: [101, 102, 103, 104, 105]
    Choose room: 103
    Room 103 booked successfully!

```python
available = [101, 102, 103, 104, 105]
booked = []
while len(booked) < len(available):
    print("Available rooms:", [r for r in available if r not in booked])
    room = int(input("Choose room: "))
    if room in booked:
        print("Room not available.")
    elif room in available:
        booked.append(room)
        print("Room", room, "booked successfully!")
    else:
        print("Invalid room number.")
print("All rooms are booked.")
```

## Create a menu-driven banking system using loops:
* Create Account
* Deposit
* Withdraw
* Check Balance
* Exit

```python
accounts = {}
while True:
    print("\n1. Create Account")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Check Balance")
    print("5. Exit")
    choice = input("Enter your choice: ")
    if choice == "1":
        name = input("Enter account name: ")
        if name in accounts:
            print("Account already exists.")
        else:
            accounts[name] = 0
            print("Account created successfully!")
    elif choice == "2":
        name = input("Enter account name: ")
        if name in accounts:
            amount = float(input("Enter amount to deposit: "))
            if amount > 0:
                accounts[name] += amount
                print("Deposited. New balance:", accounts[name])
            else:
                print("Invalid amount.")
        else:
            print("Account not found.")
    elif choice == "3":
        name = input("Enter account name: ")
        if name in accounts:
            amount = float(input("Enter amount to withdraw: "))
            if amount <= 0:
                print("Invalid amount.")
            elif amount > accounts[name]:
                print("Insufficient balance.")
            else:
                accounts[name] -= amount
                print("Withdrawn. New balance:", accounts[name])
        else:
            print("Account not found.")
    elif choice == "4":
        name = input("Enter account name: ")
        if name in accounts:
            print("Balance:", accounts[name])
        else:
            print("Account not found.")
    elif choice == "5":
        print("Thank you for using our bank!")
        break
    else:
        print("Invalid choice. Try again.")
```

## Hotel Room Booking Tracker
### Goal:

    Record bookings for rooms on each floor.

### Steps:

    Outer loop → for 3 floors.
    
    Inner loop → for 4 rooms per floor.
    
    Ask user if booked (Y/N).
    
    Count total booked and available rooms.
    
    Print summary per floor.

### Example:
    Floor 1:
    Room 1: Y
    Room 2: N

    Room 4: Y
    Floor 2:
    Room 1: Y

    → Output: Floor 1 = 2 booked, Floor 2 = 4 booked

```python
summary = []
for floor in range(1, 4):
    print("Floor", str(floor) + ":")
    booked = 0
    available = 0
    for room in range(1, 5):
        ans = input("Room " + str(room) + ": ").upper()
        if ans == "Y":
            booked += 1
        else:
            available += 1
    summary.append((floor, booked, available))
for floor, booked, available in summary:
    print("Floor", floor, "=", booked, "booked,", available, "available")
```

## Invoice Tax Calculator
### Goal:

Take multiple items in multiple invoices and calculate tax per invoice.

### Steps:

Outer loop → for 3 invoices.

Input item price, add to list.

### Example:
    Invoice 1:

## Student Result Sheet
### Goal:

    Generate report cards for each student in a class.

### Steps:

    Outer loop → 5 students.
    
    Inner loop → 4 subjects.

    If marks < 33 → print “Failed in Subject”.

    Calculate total and grade per student.

### Example:
    Student 1: [80, 90, 70, 85] → Grade: A  
    Student 2: [50, 32, 45, 40] → Failed in Subject 2

```python
for st in range(1, 6):
    print("Enter marks for Student", st, ":")
    marks = []
    failed = False
    for sub in range(1, 5):
        m = int(input("Subject " + str(sub) + ": "))
        marks.append(m)
        if m < 33:
            print("Failed in Subject", sub)
            failed = True
    total = sum(marks)
    avg = total / 4
    if failed:
        grade = "Fail"
    elif avg >= 80:
        grade = "A"
    elif avg >= 70:
        grade = "B"
    elif avg >= 60:
        grade = "C"
    elif avg >= 50:
        grade = "D"
    else:
        grade = "E"
    print("Student", st, ":", marks, "| Total:", total, "| Grade:", grade)
```

## Holiday Package Evaluator
### Goal:

Compare multiple travel packages across features like hotel, transport, food.

### Steps:

Outer loop → 3 packages.

Inner loop → 3 features.

Input rating out of 5.

Skip if rating = 0.

Print average rating per package.

### Example:
Package 1: [4, 5, 5] → Avg = 4.67  
Package 2: [3, 4, 0] → Avg = 3.5

```python
features = ["Hotel", "Transport", "Food"]
for p in range(1, 4):
    print("Package", str(p) + ":")
    ratings = []
    for f in features:
        r = int(input(f + " rating (out of 5): "))
        if r == 0:
            continue
        ratings.append(r)
    if len(ratings) > 0:
        print("Package", p, ":", ratings, "-> Avg =", round(sum(ratings) / len(ratings), 2))
    else:
        print("Package", p, ": no ratings")
```
